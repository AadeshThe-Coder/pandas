Notes from [Kaggle Learn: Pandas](https://www.kaggle.com/learn/pandas), lesson 1.

Pandas is an open-source Python library used for data manipulation and analysis.
Its two core objects are the **DataFrame** (a table) and the **Series** (a single column).

````python
import pandas as pd
````

## 1. Creating a DataFrame
A DataFrame is a table. I create one from a **dictionary**: each key becomes a column name, each list becomes that column's values.

````python
pd.DataFrame({'Yes': [50, 21], 'No': [131, 2]})
````

### Custom row labels with `index`
By default rows are labelled 0, 1, 2… I can give them names instead with `index=`.

````python
pd.DataFrame({'Yes': [50, 25],
              'No': [1, 2]},
             index=['Product A', 'Product B'])
````

## 2. Creating a Series
A Series is one column of data. I create it from a **list**.

````python
pd.Series([1, 2, 3, 4, 5])
````

### Series with an index and a name
`index=` labels each value; `name=` names the whole Series (it becomes the column name if the Series is put in a DataFrame).

````python
pd.Series([87, 89, 79, 85],
          index=['Product 1 sales', 'Product 2 sales', 'Product 3 sales', 'Product 4 sales'],
          name='Product sales')
````

## 3. Reading a CSV file
CSV = **Comma-Separated Values**: a table stored as plain text, with values separated by commas.

```python
df = pd.read_csv("data/your_file.csv")              # read a file
df = pd.read_csv("data/your_file.csv", index_col=0)  # use the first column as the index
```
After reading, always inspect it:

````python
# A small CSV written in the notebook so this cell runs anywhere.
# With a real file, use pd.read_csv("data/file.csv") instead.
from io import StringIO

csv_text = '''product,price,stock
Pen,10,120
Notebook,45,60
Bag,600,8'''

df = pd.read_csv(StringIO(csv_text))
print(df.shape)   # (rows, columns)
df.head()         # first 5 rows
````

## 4. Writing a CSV file
`to_csv` saves a DataFrame to a file. `index=False` stops pandas writing the 0, 1, 2… row labels as an extra column.

````python
df.to_csv("products.csv", index=False)
pd.read_csv("products.csv")   # read it back to check
````

## ✍️ What I learned
- A **DataFrame** is a table; a **Series** is one column.
- Dictionary → DataFrame (keys = columns). List → Series.
- `index=` sets row labels; `name=` names a Series.
- `pd.read_csv()` reads a file; `df.to_csv(..., index=False)` writes one.
- After reading, check `df.shape` and `df.head()`.
